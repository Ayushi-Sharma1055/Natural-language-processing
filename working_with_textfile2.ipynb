```python
students=["rahul\n",
          "priya\n",
          "amit\n",
          "ayushi\n"]
```

```python
#12.| searching for a word
```

```python
search_word=input("enter word to search :")

with open("student.txt","r") as file:
  data=file.read()

if search_word.lower() in data.lower():
  print("Word Found.")
else:
  print("Word Not Found.")
```

```python
#13.counting accurrences of a word
```

```python
search_word=input("enter word to search:")
with open("student.txt","r") as file:
  data=file.read()
count=data.lower().count(search_word.lower())
print("word found",count,"times")
```
