```python
#3. read all pages from the pdf
```

```python
from PyPDF2 import PdfReader
reader=PdfReader("/content/jeff108.pdf")
for page_number, page in enumerate (reader.pages,start=1):
  print(f"\n---page{page_number}---")
  text=page.extract_text()
  print(text)
```

```python
#5.search from a word in a pdf
```

```python
from PyPDF2 import PdfReader
reader=PdfReader("/content/Pollution.pdf")
search_word="and"
for page_number, page in enumerate (reader.pages,start=1):
  text=page.extract_text()
  if search_word in text:
    print(f"found '{search_word}'on page{page_number}")
```

```python
#6. count words in pdf
```

```python
from PyPDF2 import PdfReader
reader=PdfReader("content/jeff108.pdf/")
word_count=0
for page in reader.pages:
  text=page.extract_text()
if text:
  word_count+=len(text.split())
print(f"total word count:{word_count}")
```

```python
#save extracted text into a .txt file
from PyPDF2 import PdfReader
reader = PdfReader("/content/jeff108.pdf")
with open("pdf_text.txt", "w", encoding="utf-8") as file:
  for page_number,page in enumerate (reader.pages,start=1):
    text = page.extract_text()

    file.write(f"\n---Page {page_number}---\n")
    file.write(text if text else "")
print("Text extracted and saved to 'extracted_text.txt'.")
```

```python
#copy the first page into another PDF
from PyPDF2 import PdfReader, PdfWriter
reader = PdfReader("/content/jeff108.pdf") # real original pdf
writer = PdfWriter() #create new pdf
writer.add_page(reader.pages[0]) #copy first page
#save as another PDF
with open("first_page.pdf", "wb") as output:
  writer.write(output)
print("First page copied to 'first_page.pdf'.")
```
